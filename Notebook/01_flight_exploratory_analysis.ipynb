# US Aviation COVID-19 Impact Analysis
## 01 — Exploratory Data Analysis

### Research Objective
Analyze changes in US domestic flight operations between 2019 and 2020, focusing on flight volume, cancellations, delays, and airline performance.

**Tools:** PostgreSQL, Python, Pandas, SQLAlchemy

**Data Source:** US Bureau of Transportation Statistics (BTS)

```python
import pandas as pd
from getpass import getpass
from sqlalchemy import create_engine, text
from sqlalchemy.engine import URL

password = getpass("PostgreSQL password: ")

db_url = URL.create(
    "postgresql+psycopg",
    username="postgres",
    password=password,
    host="localhost",
    port=5432,
    database="aviation_covid_analysis"
)

engine = create_engine(db_url)

with engine.connect() as conn:
    print("Connected to:", conn.execute(text("SELECT current_database()")).scalar())
```

```python
query = """
SELECT
    table_name
FROM information_schema.tables
WHERE table_schema = 'public'
ORDER BY table_name;
"""

with engine.connect() as conn:
    tables = pd.read_sql(text(query), conn)

display(tables)
```

```python
query = """
SELECT '2019' AS year, COUNT(*) AS total_flights
FROM flights_2019

UNION ALL

SELECT '2020' AS year, COUNT(*) AS total_flights
FROM flights_2020;
"""

with engine.connect() as conn:
    results = pd.read_sql(text(query), conn)

display(results)
```
