```python
import pandas as pd

query = """
SELECT
    table_name
FROM information_schema.tables
WHERE table_schema = 'public'
ORDER BY table_name;
"""

display(pd.read_sql(query, engine))
```

```python
query = """
SELECT 'flights' AS table_name, COUNT(*) AS total_rows
FROM flights

UNION ALL

SELECT 'flights_2019', COUNT(*)
FROM flights_2019

UNION ALL

SELECT 'flights_2020', COUNT(*)
FROM flights_2020;
"""

display(pd.read_sql(query, engine))
```

```python
import pandas as pd
from pathlib import Path
from sqlalchemy import Date

data_folder = Path("../Data")

for year in [2019, 2020]:

    file_path = data_folder / f"flights_{year}.csv"
    table_name = f"flights_{year}_new"

    total_rows = 0

    for chunk in pd.read_csv(
        file_path,
        chunksize=10000,
        parse_dates=["fl_date"],
        low_memory=False
    ):

        chunk.columns = chunk.columns.str.lower()

        chunk.to_sql(
            name=table_name,
            con=engine,
            if_exists="replace" if total_rows == 0 else "append",
            index=False,
            chunksize=2000,
            dtype={"fl_date": Date()}
        )

        total_rows += len(chunk)

    print(f"{table_name}: {total_rows:,} rows imported")
```

```python
query = """
SELECT '2019' AS year, COUNT(*) AS total_flights
FROM flights_2019_new

UNION ALL

SELECT '2020', COUNT(*)
FROM flights_2020_new;
"""

display(pd.read_sql(query, engine))
```

```python
query = """
SELECT
    '2019' AS year,
    COUNT(*) AS total_flights,
    MIN(fl_date) AS first_date,
    MAX(fl_date) AS last_date
FROM flights_2019_new

UNION ALL

SELECT
    '2020',
    COUNT(*),
    MIN(fl_date),
    MAX(fl_date)
FROM flights_2020_new;
"""

display(pd.read_sql(query, engine))
```
